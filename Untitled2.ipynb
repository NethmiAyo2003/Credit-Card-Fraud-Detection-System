```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from imblearn.over_sampling import SMOTE

# 1. Load the Dataset
print("[+] Loading dataset...")
df = pd.read_csv("creditcard.csv")
print(f"[+] Dataset Loaded. Shape: {df.shape}")

# 2. Preprocessing & Feature Scaling
print("[+] Scaling 'Amount' and 'Time' features...")
scaler = StandardScaler()
df['scaled_amount'] = scaler.fit_transform(df['Amount'].values.reshape(-1, 1))
df['scaled_time'] = scaler.fit_transform(df['Time'].values.reshape(-1, 1))
df = df.drop(['Time', 'Amount'], axis=1)

# 3. Split Data into Train/Test Sets
X = df.drop('Class', axis=1)
y = df['Class']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# 4. Apply SMOTE to balance the dataset
print("[+] Handling class imbalance using SMOTE...")
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

# 5. Train the Model
print("[+] Training Random Forest Classifier (This may take a few minutes)...")
model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
model.fit(X_train_balanced, y_train_balanced)

# 6. Evaluate Model Performance
print("[+] Evaluating model performance...\n")
y_pred = model.predict(X_test)

print("--- Confusion Matrix ---")
print(confusion_matrix(y_test, y_pred))

print("\n--- Security Performance Metrics ---")
print(classification_report(y_test, y_pred))
```

```python
import joblib

joblib.dump(model, 'credit_card_fraud_model.pkl')
print("[+] Model saved successfully as 'credit_card_fraud_model.pkl'")
```

```python
import matplotlib.pyplot as plt


importances = model.feature_importances_
indices = np.argsort(importances)[::-1][:10]

plt.figure(figsize=(10, 6))
plt.title("Top 10 Most Critical Features for Fraud Detection")
plt.bar(range(10), importances[indices], align="center")
plt.xticks(range(10), X.columns[indices], rotation=45)
plt.tight_layout()
plt.show()
```
